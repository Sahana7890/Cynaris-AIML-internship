```python
import pandas as pd

data = {
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Delhi", "Maharashtra",
              "Punjab", "Gujarat", "Rajasthan", "Telangana", "Andhra Pradesh"],
    "Region": ["South", "South", "South", "North", "West",
               "North", "West", "North", "South", "South"],
    "Population": [67562686, 72147030, 35699443, 16787941,
                   124904071, 30141373, 63872399, 81032689,
                   39362732, 49577103],
    "Area": [191791, 130058, 38863, 1484,
             307713, 50362, 196024, 342239,
             112077, 162968]
}

df = pd.DataFrame(data)
```

```python
import matplotlib.pyplot as plt
for column in numeric_columns:

    plt.figure(figsize=(6,4))

    plt.hist(df[column], bins=5)

    plt.title(f"Distribution of {column}")

    plt.xlabel(column)

    plt.ylabel("Frequency")

    plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
top_states = df["State"].value_counts().head(10)

plt.figure(figsize=(8,4))

top_states.plot(kind="bar")

plt.title("Top 10 Category Counts")

plt.xlabel("State")

plt.ylabel("Count")

plt.show()
```
