```python
# Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set a clean style for plots
sns.set_style("whitegrid")
```

```python
# Create sample dataset
data = {
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Delhi", "Maharashtra",
              "Punjab", "Gujarat", "Rajasthan", "Telangana", "Andhra Pradesh"],
    "Region": ["South", "South", "South", "North", "West",
               "North", "West", "North", "South", "South"],
    "Population": [67562686, 72147030, 35699443, 16787941,
                   124904071, 30141373, 63872399, 81032689,
                   39362732, 49577103],
    "Area": [191791, 130058, 38863, 1484,
             307713, 50362, 196024, 342239,
             112077, 162968]
}

df = pd.DataFrame(data)
```

```python
# Display first five rows
df.head()
```

```python
# Line chart of Population
plt.figure(figsize=(10,5))

plt.plot(df["State"], df["Population"], marker="o")

plt.title("Population by State")

plt.xlabel("State")

plt.ylabel("Population")

plt.xticks(rotation=45)

plt.show()
```

```python
# Bar chart of Area
plt.figure(figsize=(10,5))

plt.bar(df["State"], df["Area"])

plt.title("Area by State")

plt.xlabel("State")

plt.ylabel("Area (sq km)")

plt.xticks(rotation=45)

plt.show()
```

```python
# Histogram of Population
plt.figure(figsize=(6,4))

plt.hist(df["Population"], bins=5)

plt.title("Population Distribution")

plt.xlabel("Population")

plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter plot of Population vs Area
plt.figure(figsize=(6,4))

plt.scatter(df["Area"], df["Population"])

plt.title("Population vs Area")

plt.xlabel("Area")

plt.ylabel("Population")

plt.show()
```

```python
# Box plot of Population
plt.figure(figsize=(6,4))

sns.boxplot(y=df["Population"])

plt.title("Population Box Plot")

plt.show()
```

```python
# Count plot of Region
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Region")

plt.title("Number of States by Region")

plt.show()
```

```python
# Correlation matrix
correlation = df.corr(numeric_only=True)

plt.figure(figsize=(5,4))

sns.heatmap(correlation, annot=True, cmap="Blues")

plt.title("Correlation Heatmap")

plt.show()
```
