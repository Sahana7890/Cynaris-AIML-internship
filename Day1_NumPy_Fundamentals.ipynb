```python
import pandas as pd

data = {
    "Student": ["Asha","Rahul","Priya","Kiran","Anil"],
    "Math": [85,78,92,88,76],
    "Science": [90,82,95,84,80],
    "English": [75,80,88,90,70]
}

df = pd.DataFrame(data)

df.to_csv("student_marks.csv", index=False)

print(df)
```

```python
math_scores = df["Math"]

print("Mean:", np.mean(math_scores))
```

```python
print("Standard Deviation:", np.std(math_scores))
```

```python
correlation = df["Math"].corr(df["Science"])

print("Correlation between Math and Science:", correlation)
```
