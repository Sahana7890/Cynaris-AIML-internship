```python
import pandas as pd
import numpy as np
```

```python
data = {
    "State": ["Karnataka", "Tamil Nadu", "Kerala", "Delhi", "Karnataka"],
    "Population": [67562686, 72147030, np.nan, 16787941, 67562686],
    "Area": [191791, 130058, 38863, 1484, 191791],
    "Region": ["South", "South", "South", "North", "South"]
}

df = pd.DataFrame(data)

df.to_csv("india_states.csv", index=False)
```
