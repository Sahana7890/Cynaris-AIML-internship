```python
import pandas as pd
import os
```

```python
data = {
    "State": [
        "Karnataka",
        "Tamil Nadu",
        "Kerala",
        "Maharashtra",
        "Delhi",
        "Telangana",
        "Andhra Pradesh",
        "Gujarat",
        "Punjab",
        "Rajasthan"
    ],

    "Region":[
        "South",
        "South",
        "South",
        "West",
        "North",
        "South",
        "South",
        "West",
        "North",
        "North"
    ],

    "Population":[
        67562686,
        72147030,
        35699443,
        124904071,
        16787941,
        39362732,
        49577103,
        63872399,
        30141373,
        81032689
    ],

    "Area":[
        191791,
        130058,
        38863,
        307713,
        1484,
        112077,
        162968,
        196024,
        50362,
        342239
    ]
}

df = pd.DataFrame(data)

df.to_csv("india_states.csv", index=False)

print("CSV Created Successfully")
```

```python
df = pd.read_csv("india_states.csv")
```

```python
# Display number of rows and columns
print(df.shape)
```

```python
# Display data types of each column
print(df.dtypes)
```

```python
# Display first 10 rows
print(df.head(10))
```

```python
# Filter states having population greater than 5 crore

filtered = df[df["Population"] > 50000000]

print(filtered)
```

```python
# Group data by Region and calculate total population

group = df.groupby("Region")["Population"].sum()

print(group)
```

```python
capital = pd.DataFrame({

    "State":[
        "Karnataka",
        "Tamil Nadu",
        "Kerala",
        "Delhi",
        "Gujarat"
    ],

    "Capital":[
        "Bengaluru",
        "Chennai",
        "Thiruvananthapuram",
        "New Delhi",
        "Gandhinagar"
    ]

})
```

```python
# Merge state and capital information

merged = pd.merge(
    df,
    capital,
    on="State",
    how="left"
)

print(merged)
```

```python
# Create pivot table showing total population by Region

pivot = pd.pivot_table(
    df,
    values="Population",
    index="Region",
    aggfunc="sum"
)

print(pivot)
```

```python
# Export cleaned dataframe

df.to_csv("cleaned_india_states.csv", index=False)
```
